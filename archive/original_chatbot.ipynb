```python
def simple_chatbot(user_query):
    # Dictionary of predefined responses
    responses = {
        "what is the total revenue for microsoft in 2024?": "The total revenue for Microsoft in 2024 was $245,122 million.",
        "what was the net income for tesla in 2023?": "Tesla’s net income in 2023 was $14,974 million.",
        "what is the total asset value for apple in 2023?": "The total asset value for Apple in 2023 was $364,980 million.",
    }

    # Clean and check the input
    user_query = user_query.lower().strip()
    return responses.get(user_query, "Sorry, I can only provide information on predefined queries.")

# Greet user
print("Hello! I'm your financial assistant. You can ask me about companies' financial data.")

# Run chatbot loop
while True:
    user_input = input("\nYou: ")
    if user_input.lower() in ['exit', 'quit', 'bye']:
        print("Bot: Goodbye! Have a financially smart day.")
        break
    response = simple_chatbot(user_input)
    print("Bot:", response)
```

```python
def keyword_chatbot(user_query):
    # Convert to lowercase for consistent matching
    query = user_query.lower()

    # Define keyword-based responses
    if "microsoft" in query and "revenue" in query:
        return "The total revenue for Microsoft in 2024 was $245,122 million."
    elif "tesla" in query and "net income" in query:
        return "Tesla’s net income in 2023 was $14,974 million."
    elif "apple" in query and "assets" in query:
        return "The total asset value for Apple in 2023 was $364,980 million."
    elif "tesla" in query and "liabilities" in query:
        return "Tesla’s total liabilities in 2023 were $77,000 million."
    elif "apple" in query and "revenue" in query:
        return "Apple's total revenue in 2023 was $383,285 million."
    elif "microsoft" in query and "assets" in query:
        return "Microsoft's total assets in 2024 were $450,000 million."
    else:
        return "Sorry, I can only answer financial queries about Microsoft, Apple, and Tesla."

# Greet user
print("Hello! I'm your financial assistant. Ask about revenue, assets, or liabilities for Microsoft, Apple, or Tesla.")

# Chat loop
while True:
    user_input = input("\nYou: ")
    if user_input.lower() in ['exit', 'quit', 'bye']:
        print("Bot: Goodbye! Stay financially informed.")
        break
    response = keyword_chatbot(user_input)
    print("Bot:", response)
```

```python
def financial_chatbot(user_query):
    query = user_query.lower().strip()

    # Recognized years and fallback if missing
    years = ['2022', '2023', '2024']
    year_in_query = next((year for year in years if year in query), None)

    # Responses mapped by (company, metric, year)
    responses = {
        ("microsoft", "revenue", "2024"): "The total revenue for Microsoft in 2024 was $245,122 million.",
        ("microsoft", "revenue", "2023"): "Microsoft's revenue in 2023 was $211,915 million.",
        ("microsoft", "assets", "2024"): "Microsoft's total assets in 2024 were $450,000 million.",

        ("apple", "revenue", "2023"): "Apple's total revenue in 2023 was $383,285 million.",
        ("apple", "assets", "2023"): "Apple's total asset value in 2023 was $364,980 million.",

        ("tesla", "net income", "2023"): "Tesla's net income in 2023 was $14,974 million.",
        ("tesla", "liabilities", "2023"): "Tesla's total liabilities in 2023 were $77,000 million.",
        ("tesla", "liabilities", "2022"): "Tesla's liabilities in 2022 were $67,000 million."
    }

    # Possible company and metric keywords
    companies = ['microsoft', 'apple', 'tesla']
    metrics = ['revenue', 'assets', 'liabilities', 'net income']

    # Identify company and metric in user query
    company_match = next((c for c in companies if c in query), None)
    metric_match = next((m for m in metrics if m in query), None)

    # Lookup and respond
    key = (company_match, metric_match, year_in_query)
    if key in responses:
        return responses[key]
    else:
        return "Sorry, I can only answer financial queries about Microsoft, Apple, and Tesla for the years 2022–2024."

# Greeting message
print("Hello! I'm your financial assistant.")
print("Ask me about revenue, assets, liabilities, or net income for Microsoft, Apple, or Tesla (2022–2024).")

# Chat loop
while True:
    user_input = input("\nYou: ")
    if user_input.lower().strip() in ['exit', 'quit', 'bye']:
        print("Bot: Goodbye! Stay financially informed.")
        break
    response = financial_chatbot(user_input)
    print("Bot:", response)
```

How It Works:
The chatbot accepts free-form user input and converts it to lowercase for consistency.

It scans the input for:

A recognized company (Microsoft, Apple, Tesla)

A financial metric (revenue, assets, liabilities, net income)

A specific year (2022, 2023, or 2024)

If all three components match a predefined combination, it returns the corresponding financial fact.

If any component is missing or not supported, it returns a default fallback message.

Limitations:
Predefined knowledge only: The chatbot cannot pull live or real-time data.

Fixed scope: It only understands financial data for 3 companies over 3 years.

No natural language processing: It cannot interpret complex grammar or synonym-based phrasing beyond the hardcoded keywords.

No learning or memory: It does not adapt or expand its responses over time.
